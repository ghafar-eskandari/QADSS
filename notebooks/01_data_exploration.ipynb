```python
import json
import pandas as pd

file_path = "../data/raw/webmelat_tsetmc_raw.json"

with open(file_path, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

df = pd.DataFrame(raw_data["closingPriceDaily"])

print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
print("Data types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
df["date"] = pd.to_datetime(
    df["dEven"].astype(str),
    format="%Y%m%d"
)

print(df[["dEven", "date"]].head())

print("\nDate range:")
print("Start:", df["date"].min())
print("End:", df["date"].max())

print("\nIs sorted ascending:", df["date"].is_monotonic_increasing)
print("Is sorted descending:", df["date"].is_monotonic_decreasing)
```

```python
df = df.sort_values("date").reset_index(drop=True)

print(df[["date", "pClosing", "pDrCotVal"]].head())
print()
print(df[["date", "pClosing", "pDrCotVal"]].tail())

print("\nIs sorted ascending:", df["date"].is_monotonic_increasing)
```

```python
price_cols = [
    "priceChange",
    "priceMin",
    "priceMax",
    "priceYesterday",
    "priceFirst",
    "pClosing",
    "pDrCotVal"
]

df[price_cols].describe().T
```

```python
check_cols = [
    "priceMin",
    "priceMax",
    "priceYesterday",
    "priceFirst",
    "pClosing",
    "pDrCotVal"
]

print("Number of zero values:")
print((df[check_cols] == 0).sum())

print("\nRows containing any zero price:")
zero_rows = df[(df[check_cols] == 0).any(axis=1)]

print("Count:", len(zero_rows))

print("\nFirst zero-price rows:")
zero_rows[["date"] + check_cols].head(10)
```

```python
for col in check_cols:
    zero_count = (df[col] == 0).sum()
    print(f"{col:20s} -> {zero_count:4d} rows ({zero_count / len(df) * 100:.2f}%)")

print("\nDate range of rows with zero price:")
zero_rows = df[(df[check_cols] == 0).any(axis=1)]

print("First:", zero_rows["date"].min())
print("Last :", zero_rows["date"].max())
```
